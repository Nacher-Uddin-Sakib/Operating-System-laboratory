```python
n = int(input("Enter how many process: "))
AT = []
BT = []

for i in range(n):
    a = int(input("Arrival Time: "))
    b = int(input("Burst Time: "))
    AT.append(a)
    BT.append(b)

print("\nPid\tAT\tBT\tCT\tTAT\tWT")

Total_TAT = 0
Total_WT = 0
time = 0

for i in range(n):
    if time < AT[i]:
        time = AT[i]
    time = time + BT[i]

    CT = time
    TAT = CT - AT[i]
    WT = TAT - BT[i]

    Total_TAT = Total_TAT + TAT
    Total_WT = Total_WT + WT

    print("P"+str(i+1), AT[i], BT[i], CT, TAT, WT, sep="\t")

print("\nAvarage TAT:", Total_TAT / n)
print("Avarage WT:", Total_WT / n)
```
