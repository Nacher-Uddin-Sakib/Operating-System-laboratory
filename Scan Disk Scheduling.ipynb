**For LEFT**

```python
req_seq = [0,13,41,50,63,89,100]
head = 46
pending = req_seq.copy ()
pending.sort()
left=[i for i in pending if i<head]
right=[i for i in pending if i>head]
seek_order=left[::-1]+right
totalseek = 0
for i in seek_order:
  totalseek+=abs(i-head)
  head=i
print ("Total seek left =",totalseek)
```

**For RIGHT**

```python
req_seq = [0,13,41,50,63,89,100]
head = 46
pending = req_seq.copy ()
pending.sort()
left=[i for i in pending if i<head]
right=[i for i in pending if i>head]
seek_order=right+[99]+left[::-1]
totalseek = 0
for i in seek_order:
  totalseek+=abs(i-head)
  head=i
print ("Total seek right =",totalseek)
```
