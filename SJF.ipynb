```python
n = int(input("Enter number of processes: "))

pid_list = []
arrival_list = []
burst_list = []

for i in range(n):
    pid = int(input("Process ID: "))
    arrival = int(input("Arrival Time: "))
    burst = int(input("Burst Time: "))

    pid_list.append(pid)
    arrival_list.append(arrival)
    burst_list.append(burst)

done = [False] * n
ct1 = [0] * n
tat1 = [0] * n
wt1 = [0] * n
time = 0

for x in range(n):
    idx = -1
    best = 999999

    for i in range(n):
        if arrival_list[i] <= time and done[i] == False:
            if burst_list[i] < best:
                best = burst_list[i]
                idx = i

    if idx == -1:
        time = time + 1
        x = x - 1
    else:
        time = time + burst_list[idx]
        ct1[idx] = time
        tat1[idx] = ct1[idx] - arrival_list[idx]
        wt1[idx] = tat1[idx] - burst_list[idx]
        done[idx] = True

remaining = burst_list.copy()
ct2 = [0] * n
tat2 = [0] * n
wt2 = [0] * n
time = 0
finished = 0

while finished < n:
    idx = -1
    best = 999999

    for i in range(n):
        if arrival_list[i] <= time and remaining[i] > 0:
            if remaining[i] < best:
                best = remaining[i]
                idx = i

    if idx == -1:
        time = time + 1
    else:
        remaining[idx] = remaining[idx] - 1
        time = time + 1

        if remaining[idx] == 0:
            ct2[idx] = time
            tat2[idx] = ct2[idx] - arrival_list[idx]
            wt2[idx] = tat2[idx] - burst_list[idx]
            finished = finished + 1

print("\nSJF (Non-Preemptive)")
print("PID\tArrival\tBurst\tCT\tTAT\tWT")

for i in range(n):
    print(pid_list[i], arrival_list[i], burst_list[i],
          ct1[i], tat1[i], wt1[i], sep="\t")

print("Average TAT =", sum(tat1) / n)
print("Average WT =", sum(wt1) / n)

print("\nSRTF (Preemptive)")
print("PID\tArrival\tBurst\tCT\tTAT\tWT")

for i in range(n):
    print(pid_list[i], arrival_list[i], burst_list[i],
          ct2[i], tat2[i], wt2[i], sep="\t")

print("Average TAT =", sum(tat2) / n)
print("Average WT =", sum(wt2) / n)
```
