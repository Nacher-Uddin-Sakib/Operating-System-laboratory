## **Priority Scheduling (Preemptive)**

```python
n = int(input("Enter number of processes: "))
pid = []
at = []
bt = []
pr = []

for i in range(n):
    pid.append(input("PID: "))
    at.append(int(input("Arrival Time: ")))
    bt.append(int(input("Burst Time: ")))
    pr.append(int(input("Priority: ")))

rem_bt = bt.copy()
ct = [0] * n
wt = [0] * n
tat = [0] * n
done = [0] * n
time = 0
count = 0

while count < n:
    idx = -1
    for i in range(n):
        if done[i] == 0 and at[i] <= time and rem_bt[i] > 0:
            if idx == -1 or pr[i] < pr[idx]:
                idx = i

    if idx == -1:
        time = time + 1
    else:
        rem_bt[idx] = rem_bt[idx] - 1
        time = time + 1
        if rem_bt[idx] == 0:
            ct[idx] = time
            tat[idx] = ct[idx] - at[idx]
            wt[idx] = tat[idx] - bt[idx]
            done[idx] = 1
            count = count + 1

print(f"{'PID':<5} {'AT':<5} {'BT':<5} {'PR':<5} {'CT':<5} {'TAT':<5} {'WT':<5}")
for i in range(n):
    print(f"{pid[i]:<5} {at[i]:<5} {bt[i]:<5} {pr[i]:<5} {ct[i]:<5} {tat[i]:<5} {wt[i]:<5}")
print("Average WT =", sum(wt) / n)
print("Average TAT =", sum(tat) / n)
```

## **Priority Scheduling (Non-Preemptive)**

```python
n = int(input("Enter number of processes: "))
pid = []
at = []
bt = []
pr = []

for i in range(n):
    pid.append(input("PID: "))
    at.append(int(input("Arrival Time: ")))
    bt.append(int(input("Burst Time: ")))
    pr.append(int(input("Priority: ")))

ct = [0] * n
wt = [0] * n
tat = [0] * n
done = [0] * n
time = 0
count = 0

while count < n:
    idx = -1
    for i in range(n):
        if done[i] == 0 and at[i] <= time:
            if idx == -1 or pr[i] < pr[idx]:
                idx = i

    if idx == -1:
        time = time + 1
    else:
        time = time + bt[idx]
        ct[idx] = time
        tat[idx] = ct[idx] - at[idx]
        wt[idx] = tat[idx] - bt[idx]
        done[idx] = 1
        count = count + 1

print(f"{'PID':<5} {'AT':<5} {'BT':<5} {'PR':<5} {'CT':<5} {'TAT':<5} {'WT':<5}")
for i in range(n):
    print(f"{pid[i]:<5} {at[i]:<5} {bt[i]:<5} {pr[i]:<5} {ct[i]:<5} {tat[i]:<5} {wt[i]:<5}")
print("Average WT =", sum(wt) / n)
print("Average TAT =", sum(tat) / n)
```
