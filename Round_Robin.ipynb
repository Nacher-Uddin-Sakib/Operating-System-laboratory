```python
n = int(input("Enter how many process: "))
at = []
bt = []
for i in range(n):
    at_val = int(input("Arrival Time: "))
    bt_val = int(input("Burst Time: "))
    at.append(at_val)
    bt.append(bt_val)
quantum = int(input("Enter Time Quantum: "))
rem_bt = list(bt)
ct = [0] * n
tat = [0] * n
wt = [0] * n
current_time = 0
completed = 0
ready_queue = []
visited = [False] * n
while completed < n:
    for i in range(n):
        if at[i] <= current_time and not visited[i]:
            ready_queue.append(i)
            visited[i] = True
    if not ready_queue:
        current_time += 1
        continue
    idx = ready_queue.pop(0)
    if rem_bt[idx] > quantum:
        rem_bt[idx] -= quantum
        current_time += quantum
    else:
        current_time += rem_bt[idx]
        rem_bt[idx] = 0
        ct[idx] = current_time
        tat[idx] = ct[idx] - at[idx]
        wt[idx] = tat[idx] - bt[idx]
        completed += 1
    for i in range(n):
        if at[i] <= current_time and not visited[i]:
            ready_queue.append(i)
            visited[i] = True
    if rem_bt[idx] > 0:
        ready_queue.append(idx)
print()
print(f"{'Pid':<8}{'AT':<8}{'BT':<8}{'CT':<8}{'TAT':<8}{'WT':<8}")
for i in range(n):
    pid_str = f"P{i + 1}"
    print(
        f"{pid_str:<8}{at[i]:<8}{bt[i]:<8}{ct[i]:<8}{tat[i]:<8}{wt[i]:<8}"
    )
avg_tat = sum(tat) / n
avg_wt = sum(wt) / n
print()
print(f"Avarage TAT: {avg_tat:.2f}")
print(f"Avarage WT: {avg_wt:.2f}")
```
